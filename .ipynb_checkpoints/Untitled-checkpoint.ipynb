```python
#python
# -*- coding: utf-8 -*-

# =====================================================================
# 기본_모형_MLR_AR12_일반LAD.py
#
# 목적:
#   기존 기본 MLR 모델에 AR(12) 항을 추가한 MLR-AR12 모델.
#
#   기존 MLR:
#       발전량 ~ dSSRD + dTSR + Hour
#
#   수정된 MLR-AR12:
#       발전량(t) ~ dSSRD(t) + dTSR(t) + Hour(t)
#                    + 발전량(t-1) + ... + 발전량(t-12)
#
#   즉, 각 시간대의 발전량을 예측할 때
#   직전 12개 관측시점의 실제 발전량을 추가적인 설명변수로 사용한다.
#
#   IMPORTANT:
#   AR lag 12개는 사용자가 지정한 기존 방식대로
#       previous_12_values[::-1]
#   을 사용하여 시간 순서를 거꾸로 저장한다.
#
#   따라서 AR feature의 열 순서는:
#       AR_LAG_1 = 가장 최근 관측값(t-1)
#       AR_LAG_2 = 그 이전 관측값(t-2)
#       ...
#       AR_LAG_12 = 12번째 이전 관측값(t-12)
#
#   일반 LAD:
#       학습 과정에서는 예측값의 [0,1] 범위를 강제하지 않는다.
#       즉 beta에는 별도의 [0,1] 제약을 걸지 않는다.
#
#   테스트 예측 후:
#       prediction < 0  -> 0
#       prediction > 1  -> 1
#       로 사후 clipping 한다.
#
# 데이터:
#   merged_for_simulation_z01.csv (Zone1)
#
# 구간:
#   학습: 2012-11-28 ~ 2013-09-23
#   테스트: 2013-09-24 ~ 2014-01-01
#
# 주의:
#   여기서 AR(12)의 lag는 daylight 데이터의 시간 순서 기준이다.
#   즉 필터링된 9~20시 관측값을 하나의 시계열로 보고
#   직전 12개 관측값을 lag로 사용한다.
# =====================================================================


import os
import numpy as np
import pandas as pd
from scipy import sparse
from scipy.optimize import linprog


# =====================================================================
# 0. 설정값
# =====================================================================

# 현재 작업 디렉토리를 기본 경로로 사용
BASE_DIR = os.getcwd()

# 읽어올 데이터 파일
MERGED_FILE = os.path.join(
    BASE_DIR,
    "merged_for_simulation_z01.csv"
)

# ---------------------------------------------------------------------
# 낮 시간대 설정
# ---------------------------------------------------------------------

LOCAL_HOUR_START = 9
LOCAL_HOUR_END = 21

# ---------------------------------------------------------------------
# 학습 / 테스트 기간
# ---------------------------------------------------------------------

TRAIN_START = pd.Timestamp("2012-11-28")
TRAIN_END = pd.Timestamp("2013-09-23")

TEST_START = pd.Timestamp("2013-09-24")
TEST_END = pd.Timestamp("2014-01-01")

# ---------------------------------------------------------------------
# AR 설정
# ---------------------------------------------------------------------

AR_ORDER = 12

# AR(12)의 첫 번째 lag를 만들기 위해
# 학습 시작일 바로 전날의 데이터를 사용한다.
HISTORY_DATE = pd.Timestamp("2012-11-27")

# ---------------------------------------------------------------------
# 시장 / 수익 계산 설정
# ---------------------------------------------------------------------

CAPACITY_MW = 30.0
DURATION_HOURS = 1.0
PENALTY_RATE = 0.5

# ---------------------------------------------------------------------
# 논문 결과 비교용
# ---------------------------------------------------------------------

PAPER_NRMSE = 21.76
PAPER_GAP = 12.59

# 기존 bounded LAD MLR 결과
BOUNDED_LAD_NRMSE = 32.183847
BOUNDED_LAD_GAP = 2.745470


# =====================================================================
# 1. 데이터 읽기 + Sydney 현지시간 낮 시간대만 남기기
# =====================================================================

raw_table = pd.read_csv(MERGED_FILE)

# local_date를 날짜 타입으로 변환
raw_table["local_date"] = pd.to_datetime(
    raw_table["local_date"]
)

# 9시 이상 21시 미만만 선택
is_daylight = (
    (raw_table["local_hour"] >= LOCAL_HOUR_START)
    &
    (raw_table["local_hour"] < LOCAL_HOUR_END)
)

# daylight 데이터만 남김
daylight_table = raw_table[is_daylight].copy()

# 9시를 0번으로 다시 번호 매김
# 9 -> 0
# 10 -> 1
# ...
# 20 -> 11
daylight_table["hour_idx"] = (
    daylight_table["local_hour"]
    - LOCAL_HOUR_START
)


# =====================================================================
# 2. history / train / test 구간으로 분리
# =====================================================================

# ---------------------------------------------------------------------
# history
# ---------------------------------------------------------------------

is_history_date = (
    daylight_table["local_date"] == HISTORY_DATE
)

history_rows = daylight_table[
    is_history_date
].copy()

history_rows = history_rows.sort_values(
    ["local_date", "hour_idx"]
)

# ---------------------------------------------------------------------
# train
# ---------------------------------------------------------------------

is_train_date = (
    (daylight_table["local_date"] >= TRAIN_START)
    &
    (daylight_table["local_date"] <= TRAIN_END)
)

train_rows = daylight_table[
    is_train_date
].copy()

train_rows = train_rows.sort_values(
    ["local_date", "hour_idx"]
)

# ---------------------------------------------------------------------
# test
# ---------------------------------------------------------------------

is_test_date = (
    (daylight_table["local_date"] >= TEST_START)
    &
    (daylight_table["local_date"] <= TEST_END)
)

test_rows = daylight_table[
    is_test_date
].copy()

test_rows = test_rows.sort_values(
    ["local_date", "hour_idx"]
)

print(
    "History:",
    HISTORY_DATE.date(),
    "행 수:",
    len(history_rows)
)

print(
    "학습 구간:",
    TRAIN_START.date(),
    "~",
    TRAIN_END.date(),
    "행 수:",
    len(train_rows)
)

print(
    "테스트 구간:",
    TEST_START.date(),
    "~",
    TEST_END.date(),
    "행 수:",
    len(test_rows)
)


# =====================================================================
# 3. 학습 / 테스트용 1차원 배열 만들기
# =====================================================================

# ---------------------------------------------------------------------
# 3-1. History
# ---------------------------------------------------------------------

n_history_obs = len(history_rows)

history_solar = np.zeros(
    n_history_obs
)

history_counter = 0

for _, one_row in history_rows.iterrows():

    history_solar[history_counter] = (
        one_row["solar_power"]
    )

    history_counter = history_counter + 1


# ---------------------------------------------------------------------
# 3-2. Train
# ---------------------------------------------------------------------

n_train_obs = len(train_rows)

train_solar = np.zeros(
    n_train_obs
)

train_dssrd = np.zeros(
    n_train_obs
)

train_dtsr = np.zeros(
    n_train_obs
)

train_hour = np.zeros(
    n_train_obs
)

train_counter = 0

for _, one_row in train_rows.iterrows():

    train_solar[train_counter] = (
        one_row["solar_power"]
    )

    train_dssrd[train_counter] = (
        one_row["dssrd"]
    )

    train_dtsr[train_counter] = (
        one_row["dtsr"]
    )

    train_hour[train_counter] = (
        one_row["hour_idx"]
    )

    train_counter = train_counter + 1


# ---------------------------------------------------------------------
# 3-3. Test
# ---------------------------------------------------------------------

n_test_obs = len(test_rows)

test_solar = np.zeros(
    n_test_obs
)

test_dssrd = np.zeros(
    n_test_obs
)

test_dtsr = np.zeros(
    n_test_obs
)

test_hour = np.zeros(
    n_test_obs
)

test_da_price = np.zeros(
    n_test_obs
)

test_rt_price = np.zeros(
    n_test_obs
)

test_counter = 0

for _, one_row in test_rows.iterrows():

    test_solar[test_counter] = (
        one_row["solar_power"]
    )

    test_dssrd[test_counter] = (
        one_row["dssrd"]
    )

    test_dtsr[test_counter] = (
        one_row["dtsr"]
    )

    test_hour[test_counter] = (
        one_row["hour_idx"]
    )

    test_da_price[test_counter] = (
        one_row["da_price"]
    )

    test_rt_price[test_counter] = (
        one_row["rt_price"]
    )

    test_counter = test_counter + 1


# =====================================================================
# 4. AR(12) 입력변수 만들기
#
# 핵심 수정 부분
#
# 기존 MLR에는 AR 항이 없었음.
#
# 이제 각 관측시점 t에 대해서:
#
#     AR_LAG_1  = y(t-1)
#     AR_LAG_2  = y(t-2)
#     ...
#     AR_LAG_12 = y(t-12)
#
# 를 사용한다.
#
# 사용자가 요청한 대로
#
#     previous_12_values[::-1]
#
# 을 사용하여 12개 값을 뒤집어서 저장한다.
# =====================================================================

# ---------------------------------------------------------------------
# 4-1. History + Train을 하나의 시간순 시계열로 연결
# ---------------------------------------------------------------------

history_and_train_solar = np.concatenate(
    [
        history_solar,
        train_solar
    ]
)

# ---------------------------------------------------------------------
# 4-2. Train AR(12) feature 생성
# ---------------------------------------------------------------------

ar_lag_features_train = np.zeros(
    (n_train_obs, AR_ORDER)
)

for i in range(n_train_obs):

    # 현재 train 관측값의 위치
    current_position = n_history_obs + i

    # 현재 시점보다 앞에 있는 12개 발전량을 가져온다.
    #
    # 예:
    #
    # [t-12, t-11, ..., t-2, t-1]
    #
    previous_12_values = (
        history_and_train_solar[
            current_position - AR_ORDER:
            current_position
        ]
    )

    # 사용자가 요청한 방식 그대로
    #
    # [t-12, ..., t-1]
    #        ↓ [::-1]
    # [t-1, ..., t-12]
    #
    ar_lag_features_train[i] = (
        previous_12_values[::-1]
    )


# ---------------------------------------------------------------------
# 4-3. AR feature 확인
# ---------------------------------------------------------------------

print()
print("=== AR(12) 학습 입력 확인 ===")
print(
    "AR feature shape:",
    ar_lag_features_train.shape
)

print(
    "첫 번째 학습 관측치의 AR 입력:",
    ar_lag_features_train[0]
)

print(
    "첫 번째 학습 관측치의 의미:",
    "[t-1, t-2, ..., t-12]"
)


# =====================================================================
# 5. MLR + AR(12) 회귀 입력행렬 X 만들기
#
# 기존 MLR:
#
#     [1, dSSRD, dTSR, Hour]
#
# 수정된 MLR-AR12:
#
#     [1,
#      dSSRD,
#      dTSR,
#      Hour,
#      AR_LAG_1,
#      ...
#      AR_LAG_12]
#
# 총 16개 변수
# =====================================================================

N_BASE_FEATURES = 4

N_AR_FEATURES = AR_ORDER

n_features = (
    N_BASE_FEATURES
    +
    N_AR_FEATURES
)

# 총 16개
# 1 + 3개 MLR 변수 + 12개 AR 변수
#
# 실제로는:
# intercept = 1
# dSSRD     = 1
# dTSR      = 1
# Hour      = 1
# AR        = 12
#
# = 16

print()
print(
    "총 회귀 변수 개수:",
    n_features
)


# ---------------------------------------------------------------------
# 5-1. Train X
# ---------------------------------------------------------------------

X_train = np.zeros(
    (n_train_obs, n_features)
)

for i in range(n_train_obs):

    # ---------------------------------
    # 기존 MLR 변수
    # ---------------------------------

    X_train[i, 0] = 1.0

    X_train[i, 1] = (
        train_dssrd[i]
    )

    X_train[i, 2] = (
        train_dtsr[i]
    )

    X_train[i, 3] = (
        train_hour[i]
    )

    # ---------------------------------
    # AR(12) 변수
    # ---------------------------------

    X_train[
        i,
        4:4 + AR_ORDER
    ] = ar_lag_features_train[i]


# ---------------------------------------------------------------------
# 5-2. Test AR(12) feature 생성
#
# 테스트 첫 관측값부터 직전 12개 실제 발전량을 사용한다.
#
# 테스트 기간에서는 실제 발전량을 사용하여
# rolling one-step-ahead 방식으로 AR 입력을 구성한다.
# ---------------------------------------------------------------------

# History + Train 전체를 이용하여
# 테스트 첫 시점 이전의 12개 관측값을 확보한다.
#
# 마지막 12개:
#
# [t-12, ..., t-1]
#
# 을 가져온다.

previous_12_for_test = (
    history_and_train_solar[-AR_ORDER:]
)

# 테스트 전체의 AR 입력을 저장
ar_lag_features_test = np.zeros(
    (n_test_obs, AR_ORDER)
)

# 테스트 시점마다 rolling
for i in range(n_test_obs):

    # 현재 시점 바로 직전 12개
    previous_12_values = (
        previous_12_for_test.copy()
    )

    # 사용자가 지정한 방식 그대로
    #
    # [t-12, ..., t-1]
    #       ↓
    # [t-1, ..., t-12]
    #
    ar_lag_features_test[i] = (
        previous_12_values[::-1]
    )

    # ---------------------------------------------------------------
    # 다음 테스트 시점으로 이동
    #
    # 오늘의 실제 발전량을 다음 시점의 가장 최근 lag로 추가
    # ---------------------------------------------------------------

    previous_12_for_test = np.concatenate(
        [
            previous_12_for_test[1:],
            [test_solar[i]]
        ]
    )


# ---------------------------------------------------------------------
# 5-3. Test X
# ---------------------------------------------------------------------

X_test = np.zeros(
    (n_test_obs, n_features)
)

for i in range(n_test_obs):

    # ---------------------------------
    # 기존 MLR 변수
    # ---------------------------------

    X_test[i, 0] = 1.0

    X_test[i, 1] = (
        test_dssrd[i]
    )

    X_test[i, 2] = (
        test_dtsr[i]
    )

    X_test[i, 3] = (
        test_hour[i]
    )

    # ---------------------------------
    # AR(12) 변수
    # ---------------------------------

    X_test[
        i,
        4:4 + AR_ORDER
    ] = ar_lag_features_test[i]


# =====================================================================
# 6. 일반 LAD 회귀
#
# 목적:
#
#     min_beta Σ |y_i - X_i beta|
#
# [0,1] 예측값 제약은 학습 과정에서 사용하지 않는다.
#
# 따라서:
#
#     beta = 자유
#     u >= 0
#
# 만 사용한다.
# =====================================================================

X_sparse = sparse.csr_matrix(
    X_train
)

identity_matrix = sparse.eye(
    n_train_obs,
    format="csr"
)


# ---------------------------------------------------------------------
# 절대오차를 LP 형태로 표현
#
# X beta - u <= y
# -X beta - u <= -y
#
# 두 식을 합치면
#
#     u >= |X beta - y|
#
# 가 된다.
# ---------------------------------------------------------------------

constraint_block_1 = sparse.hstack(
    [
        X_sparse,
        -identity_matrix
    ]
)

constraint_block_2 = sparse.hstack(
    [
        -X_sparse,
        -identity_matrix
    ]
)


# 두 제약식을 합침
all_constraints = sparse.vstack(
    [
        constraint_block_1,
        constraint_block_2
    ],
    format="csr"
)


# ---------------------------------------------------------------------
# 제약식 우변
# ---------------------------------------------------------------------

constraint_limits = np.concatenate(
    [
        train_solar,
        -train_solar
    ]
)


# ---------------------------------------------------------------------
# 목적함수
#
# beta에는 비용을 주지 않고
# u의 평균을 최소화한다.
# ---------------------------------------------------------------------

objective_coefficients = np.concatenate(
    [
        np.zeros(n_features),
        np.ones(n_train_obs) / n_train_obs
    ]
)


# ---------------------------------------------------------------------
# 변수 범위
#
# beta:
#     제한 없음
#
# u:
#     0 이상
# ---------------------------------------------------------------------

variable_bounds = (
    [(None, None)] * n_features
    +
    [(0.0, None)] * n_train_obs
)


# ---------------------------------------------------------------------
# LP 실행
# ---------------------------------------------------------------------

lp_result = linprog(
    objective_coefficients,
    A_ub=all_constraints,
    b_ub=constraint_limits,
    bounds=variable_bounds,
    method="highs"
)


# ---------------------------------------------------------------------
# 회귀계수 추출
# ---------------------------------------------------------------------

mlr_ar12_coefficients = (
    lp_result.x[:n_features]
)


print()
print(
    "MLR-AR12 회귀 완료 "
    "(성공 여부:",
    lp_result.success,
    ")"
)

print()
print("=== MLR-AR12 회귀계수 ===")

print(
    "Intercept:",
    mlr_ar12_coefficients[0]
)

print(
    "dSSRD:",
    mlr_ar12_coefficients[1]
)

print(
    "dTSR:",
    mlr_ar12_coefficients[2]
)

print(
    "Hour:",
    mlr_ar12_coefficients[3]
)

for lag in range(AR_ORDER):

    print(
        f"AR_LAG_{lag + 1}:",
        mlr_ar12_coefficients[4 + lag]
    )


# =====================================================================
# 7. 테스트 구간 예측
#
# MLR + AR(12)
#
# 예측식:
#
# y_hat(t) =
#
# beta_0
# + beta_1*dSSRD(t)
# + beta_2*dTSR(t)
# + beta_3*Hour(t)
# + beta_4*y(t-1)
# + ...
# + beta_15*y(t-12)
#
# 학습에서는 일반 LAD.
#
# 예측 후에만 [0,1] clipping.
# =====================================================================

test_forecast = np.zeros(
    n_test_obs
)

for i in range(n_test_obs):

    # ---------------------------------------------------------------
    # MLR-AR12 입력을 이용하여 raw prediction 계산
    # ---------------------------------------------------------------

    raw_prediction = np.dot(
        mlr_ar12_coefficients,
        X_test[i]
    )

    # ---------------------------------------------------------------
    # 사후 clipping
    #
    # 학습 중에는 [0,1] 제약이 없었지만
    # 실제 태양광 발전량 예측값은 [0,1] 범위여야 하므로
    # 예측이 끝난 뒤에만 잘라낸다.
    # ---------------------------------------------------------------

    clipped_prediction = min(
        max(raw_prediction, 0.0),
        1.0
    )

    test_forecast[i] = (
        clipped_prediction
    )


# =====================================================================
# 8. nRMSE 계산
# =====================================================================

sum_of_squared_error = 0.0

for i in range(n_test_obs):

    error_i = (
        test_solar[i]
        -
        test_forecast[i]
    )

    sum_of_squared_error = (
        sum_of_squared_error
        +
        error_i * error_i
    )


mean_squared_error = (
    sum_of_squared_error
    /
    n_test_obs
)

rmse_value = (
    mean_squared_error ** 0.5
)


# ---------------------------------------------------------------------
# 실제 발전량 평균
# ---------------------------------------------------------------------

sum_of_actual = 0.0

for i in range(n_test_obs):

    sum_of_actual = (
        sum_of_actual
        +
        test_solar[i]
    )


average_actual = (
    sum_of_actual
    /
    n_test_obs
)


# ---------------------------------------------------------------------
# nRMSE
# ---------------------------------------------------------------------

nrmse_percent = (
    100.0
    *
    rmse_value
    /
    average_actual
)


# =====================================================================
# 9. Optimality gap 계산
#
# 논문 Eq.(1a) 이익함수 3항
# + Eq.(13) oracle {0,S}
# =====================================================================

sum_of_realized_profit = 0.0

sum_of_oracle_profit = 0.0


for i in range(n_test_obs):

    # ---------------------------------------------------------------
    # 현재 시간의 값
    # ---------------------------------------------------------------

    actual_i = (
        test_solar[i]
    )

    commitment_i = (
        test_forecast[i]
    )

    da_i = (
        test_da_price[i]
    )

    rt_i = (
        test_rt_price[i]
    )

    # 부족분 벌금 단가
    penalty_cost_i = (
        PENALTY_RATE
        *
        da_i
    )


    # ---------------------------------------------------------------
    # 실제 약정 기반 수익
    # ---------------------------------------------------------------

    mismatch_i = (
        actual_i
        -
        commitment_i
    )

    surplus_i = max(
        mismatch_i,
        0.0
    )

    shortage_i = max(
        -mismatch_i,
        0.0
    )


    realized_profit_i = (
        CAPACITY_MW
        *
        DURATION_HOURS
        *
        (
            da_i * commitment_i
            +
            rt_i * surplus_i
            -
            penalty_cost_i * shortage_i
        )
    )

    sum_of_realized_profit = (
        sum_of_realized_profit
        +
        realized_profit_i
    )


    # ---------------------------------------------------------------
    # Oracle
    #
    # 약정 = 0
    # 약정 = 실제 발전량
    #
    # 두 경우 중 더 큰 이익을 선택
    # ---------------------------------------------------------------

    profit_if_commit_zero = (
        CAPACITY_MW
        *
        DURATION_HOURS
        *
        (
            rt_i
            *
            actual_i
        )
    )

    profit_if_commit_actual = (
        CAPACITY_MW
        *
        DURATION_HOURS
        *
        (
            da_i
            *
            actual_i
        )
    )

    oracle_profit_i = max(
        profit_if_commit_zero,
        profit_if_commit_actual
    )

    sum_of_oracle_profit = (
        sum_of_oracle_profit
        +
        oracle_profit_i
    )


# ---------------------------------------------------------------------
# Optimality gap
# ---------------------------------------------------------------------

optimality_gap_percent = (
    100.0
    *
    (
        sum_of_oracle_profit
        -
        sum_of_realized_profit
    )
    /
    sum_of_oracle_profit
)


# =====================================================================
# 10. 결과 출력
# =====================================================================

print()

print(
    "============================================================"
)

print(
    "=== 기본 모형 MLR-AR(12) "
    "(일반 LAD, 제약 없음) ==="
)

print(
    "============================================================"
)

print()

print(
    f"nRMSE          = "
    f"{nrmse_percent:.6f} %"
)

print(
    f"optimality gap = "
    f"{optimality_gap_percent:.6f} %"
)

print()

print(
    "논문 기본 MLR 참고값:"
)

print(
    f"nRMSE          = "
    f"{PAPER_NRMSE:.2f} %"
)

print(
    f"optimality gap = "
    f"{PAPER_GAP:.2f} %"
)

print()

print(
    "기존 bounded LAD MLR:"
)

print(
    f"nRMSE          = "
    f"{BOUNDED_LAD_NRMSE:.6f} %"
)

print(
    f"optimality gap = "
    f"{BOUNDED_LAD_GAP:.6f} %"
)

print()


# =====================================================================
# 11. 최종 비교표
# =====================================================================

print(
    "| 회귀 방식 | nRMSE | optimality gap |"
)

print(
    "|---|---:|---:|"
)

print(
    f"| 논문 기본 MLR (참고) | "
    f"{PAPER_NRMSE:.2f}% | "
    f"{PAPER_GAP:.2f}% |"
)

print(
    f"| bounded LAD MLR | "
    f"{BOUNDED_LAD_NRMSE:.6f}% | "
    f"{BOUNDED_LAD_GAP:.6f}% |"
)

print(
    f"| **MLR + AR(12), 일반 LAD** | "
    f"**{nrmse_percent:.6f}%** | "
    f"**{optimality_gap_percent:.6f}%** |"
)


# =====================================================================
# 12. AR(12) 입력 구조 최종 확인
# =====================================================================

print()
print(
    "============================================================"
)

print(
    "=== AR(12) 입력 구조 확인 ==="
)

print(
    "============================================================"
)

print(
    "X_train shape:",
    X_train.shape
)

print(
    "X_test shape:",
    X_test.shape
)

print()

print(
    "각 행의 변수 순서:"
)

print(
    "0  : Intercept"
)

print(
    "1  : dSSRD"
)

print(
    "2  : dTSR"
)

print(
    "3  : Hour"
)

for lag in range(AR_ORDER):

    print(
        f"{4 + lag:2d} : "
        f"AR_LAG_{lag + 1} = "
        f"y(t-{lag + 1})"
    )

print()

print(
    "AR 입력은 previous_12_values[::-1] 방식으로"
)

print(
    "[t-12, ..., t-2, t-1]"
    " -> "
    "[t-1, t-2, ..., t-12]"
)

print(
    "순서로 저장됩니다."
)



```
